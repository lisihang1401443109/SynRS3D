```python
train_file = '/mnt/synrs3d/SynRS3D/data/OEM/test.txt'
base = '/mnt/synrs3d/SynRS3D/data/OEM/'
```

```python
noss, noimg = 0, 0
noboth = 0
with open(train_file, 'r') as f:
    for line in f:
        fn = line.strip()
        
        # print(os.path.isfile(base+'opt/'+fn), (base+'opt/'+fn))
        # check if both exists
        if os.path.isfile(base+'opt/'+fn) or os.path.isfile(base+'gt_ss_mask/'+fn):
            if os.path.isfile(base+'opt/'+fn):
                noss += 1
            else:
                noimg += 1
        else:
            noboth += 1
        
print(f'{noss}, {noimg}, {noboth} not found')
```

```python
import os

base = '/mnt/synrs3d/SynRS3D/data/OEM/'
opt_dir = os.path.join(base, 'opt')
gt_dir = os.path.join(base, 'gt_ss_mask')

# List all .tif files in both directories
opt_files = {f for f in os.listdir(opt_dir) if f.endswith('.tif')}
gt_files = {f for f in os.listdir(gt_dir) if f.endswith('.tif')}

# Find common files
common_files = sorted(opt_files & gt_files)

# print(f"Found {len(common_files)} matching .tif files:")
# for fname in common_files:
#     print(fname)

```

```python
common_files
```

```python
import random

random.seed(42)  # Optional: for reproducibility
sampled_files = random.sample(common_files, 1000)
```

```python
output_file = os.path.join(base, 'new_test.txt')

# Write each filename to a new line in the output file
with open(output_file, 'w') as f:
    for fname in sampled_files:
        f.write(f"{fname}\n")
```
