```python
import numpy as np

def be(x):
    return -x * np.log2(x) - (1-x) * np.log2(1-x)
```

```python
be(2/3)
```

```python
#get entropy of a list of probabilities
def get_entropy(probs):
    return -np.sum(probs * np.log(probs))
    
get_entropy([3/16, 3/16, 3/16, 3/16, 4/16])
```

```python
np.log(16) - get_entropy([3/16, 3/16, 3/16, 3/16, 4/16])
```
